```python
def radon_scratch(image, theta):
    size = image.shape[0]
    sinogram_pix = np.zeros((size, len(theta)), dtype=np.float16)
    rot_pix = np.zeros((size, size))
    center = size//2
    for i, angle in enumerate(theta):
        c, s = np.cos(np.deg2rad(angle)), np.sin(np.deg2rad(angle))
        R = np.array([[c, s], [-s, c]])
        for j in range(size):
            for k in range(size):
                x, y = np.array([j, k]) - center
                x_rot, y_rot = np.dot(R, np.array([x, y]))
                x_rot_idx, y_rot_idx = np.uint8(x_rot + center), np.uint8((y_rot + center))
                if x_rot_idx >= 0 and x_rot_idx < size and y_rot_idx >= 0 and y_rot_idx < size:
                    rot_pix[x_rot_idx, y_rot_idx] = image[j, k]
                sinogram_pix[:, i] = rot_pix.sum(axis=0)
    return sinogram_pix
```

```python
# Which step takes the longest? -> instrumented copy of radon_scratch (identical logic, timers around each step),
# run on 10 angles only (the cost is the same for every angle).
def radon_scratch_profiled(image, theta):
    size = image.shape[0]
    sinogram_pix = np.zeros((size, len(theta)), dtype=np.float16)
    rot_pix = np.zeros((size, size)); center = size // 2
    T = dict(rotation_matrix=0., coordinates_and_dot=0., cast_and_scatter=0., sum_inside_inner_loop=0.)
    pc = time.perf_counter
    for i, angle in enumerate(theta):
        t = pc(); c, s = np.cos(np.deg2rad(angle)), np.sin(np.deg2rad(angle)); R = np.array([[c, s], [-s, c]])
        T["rotation_matrix"] += pc() - t
        for j in range(size):
            for k in range(size):
                t = pc()
                x, y = np.array([j, k]) - center
                x_rot, y_rot = np.dot(R, np.array([x, y]))
                t1 = pc(); T["coordinates_and_dot"] += t1 - t
                x_rot_idx, y_rot_idx = np.uint8(x_rot + center), np.uint8((y_rot + center))
                if x_rot_idx >= 0 and x_rot_idx < size and y_rot_idx >= 0 and y_rot_idx < size:
                    rot_pix[x_rot_idx, y_rot_idx] = image[j, k]
                t2 = pc(); T["cast_and_scatter"] += t2 - t1
                sinogram_pix[:, i] = rot_pix.sum(axis=0)
                T["sum_inside_inner_loop"] += pc() - t2
    return sinogram_pix, T

_, T = radon_scratch_profiled(GT, THETA[:10])
tot = sum(T.values())
plt.figure(figsize=(7, 2.8))
plt.barh(list(T.keys()), [100 * v / tot for v in T.values()])
plt.xlabel("% of the run time"); plt.title("radon_scratch: share of each step (10 angles)"); plt.grid(alpha=.3)
plt.tight_layout(); plt.show()
for k, v in T.items():
    print(f"{k:<24s}: {100*v/tot:5.1f} %")

# Moving the projection sum OUT of the two pixel loops gives exactly the same sinogram (only the last value was kept):
def radon_scratch_sum_out(image, theta):
    size = image.shape[0]
    sinogram_pix = np.zeros((size, len(theta)), dtype=np.float16)
    rot_pix = np.zeros((size, size)); center = size // 2
    for i, angle in enumerate(theta):
        c, s = np.cos(np.deg2rad(angle)), np.sin(np.deg2rad(angle)); R = np.array([[c, s], [-s, c]])
        for j in range(size):
            for k in range(size):
                x, y = np.array([j, k]) - center
                x_rot, y_rot = np.dot(R, np.array([x, y]))
                x_rot_idx, y_rot_idx = np.uint8(x_rot + center), np.uint8((y_rot + center))
                if x_rot_idx >= 0 and x_rot_idx < size and y_rot_idx >= 0 and y_rot_idx < size:
                    rot_pix[x_rot_idx, y_rot_idx] = image[j, k]
        sinogram_pix[:, i] = rot_pix.sum(axis=0)          # <- once per angle
    return sinogram_pix
t0 = time.perf_counter(); s_out = radon_scratch_sum_out(GT, THETA); t_out = time.perf_counter() - t0
print(f"\nsum moved out of the pixel loops: {t_out*1e3:.0f} ms (vs {t_scratch*1e3:.0f} ms), "
      f"identical output: {np.array_equal(s_out, sino_scratch)}")
```

```python
#  A corrected, vectorised "from scratch" Radon (same geometry as skimage)
# inverse mapping (for every output pixel, fetch the source position), bilinear interpolation written by hand,
# float64, rotation centre = size//2 exactly as skimage
def radon_scratch_v2(image, theta):
    size = image.shape[0]; center = size // 2
    sino = np.zeros((size, len(theta)))
    r, c = np.indices((size, size)).astype(float)
    x, y = c - center, r - center                         # output coordinates (x = column, y = row)
    for i, a in enumerate(np.deg2rad(theta)):
        ca, sa = np.cos(a), np.sin(a)
        xs = ca * x + sa * y + center                     # source column  (skimage matrix [[c, s], [-s, c]])
        ys = -sa * x + ca * y + center                    # source row
        x0, y0 = np.floor(xs).astype(int), np.floor(ys).astype(int)
        dx, dy = xs - x0, ys - y0
        out = np.zeros((size, size))
        for oy, ox, wgt in [(0, 0, (1-dx)*(1-dy)), (0, 1, dx*(1-dy)), (1, 0, (1-dx)*dy), (1, 1, dx*dy)]:
            yi, xi = y0 + oy, x0 + ox
            ok = (yi >= 0) & (yi < size) & (xi >= 0) & (xi < size)
            out[ok] += wgt[ok] * image[yi[ok], xi[ok]]
        sino[:, i] = out.sum(axis=0)
    return sino

t_v2 = timeit.timeit(lambda: radon_scratch_v2(GT, THETA), number=3) / 3
sino_v2 = radon_scratch_v2(GT, THETA)
print(f"radon_scratch_v2: {t_v2*1e3:.1f} ms  ({t_v2/t_sk:.1f}x skimage)")
report("sinogram v2 vs skimage", sino_sk, sino_v2)
```

**INTERPRETATIONS**

**1) Execution time** (`time.perf_counter` / `timeit`; exact values printed above, they vary slightly between runs). `skimage.radon` takes a few tens of ms (≈ 35–65 ms depending on the run), `radon_scratch` ≈ 30 s, i.e. **~500–800× slower**. skimage rotates the whole image at once with `warp` (compiled, vectorised bilinear interpolation) and sums the columns, whereas `radon_scratch` loops in Python over 200 angles × 10 000 pixels = 2·10⁶ iterations, each creating small numpy arrays. The instrumented copy shows that the **longest step is `sinogram_pix[:, i] = rot_pix.sum(axis=0)`, wrongly placed inside the double pixel loop (~50 % of the time)**: a full 100×100 sum is recomputed 10 000 times per angle although only the last one is kept. Moving it after the loops gives an identical sinogram 2–4× faster. The rest is Python overhead of the per-pixel coordinate computation (`np.array` + `np.dot`, ~37 %) and of the cast/scatter (~13 %).

**2) Sinograms.** Both sinograms have the same global shape and the same orientation (rotation direction matches skimage), but `radon_scratch` is noisy/speckled: SNR = 22.5 dB w.r.t. skimage (MSE 1.08, max |diff| 8.4 for a max of 25.7), with the largest errors on the skull trajectories. Causes, all in the code: (i) **forward mapping** (each input pixel is sent to a rounded output position): some output pixels receive nothing (holes) and others receive several values (overwrites), so the mass per projection fluctuates (1219–1240 instead of 1231.4); (ii) `rot_pix` is **never reset** between angles, so holes keep stale values from the previous angle; (iii) `np.uint8(...)` **truncates** instead of rounding (half-pixel shift, largest at the sharp skull edges) and negative indices **wrap around** (−1.5 → 255), only hidden by the bound test by luck; (iv) the sinogram is stored in **float16** (3 significant digits, step 0.016 around 25). The corrected vectorised version `radon_scratch_v2` (inverse mapping + hand-written bilinear interpolation, float64, centre N//2) reproduces skimage to MSE 2.4e-28 in ≈ 0.1 s.

**3) Reconstructions (`iradon`).** iradon(skimage sinogram): MSE 1.23e-3, SNR 16.2 dB w.r.t. GT. iradon(radon_scratch): MSE 2.69e-3, SNR 12.8 dB, i.e. 2.2× more error; between the two reconstructions, SNR 18.9 dB. The sinogram errors are high-frequency and inconsistent between angles, so the ramp filter amplifies them into fine streaks/noise spread over the image; the systematic truncation shift blurs the edges. With `radon_scratch_v2` the reconstruction is identical to skimage (16.2 dB).

**4) Profiles.** Both profiles follow the GT plateaus (skull 1.0, brain 0.2, ellipses 0.3–0.4). The scratch reconstruction oscillates around the GT in the homogeneous regions and the edges are less steep; the small ellipses on row 80 are less contrasted. Profile MSE: vertical 6.0e-4 (skimage) vs 1.8e-3 (scratch); horizontal 6.7e-5 vs 3.1e-4.

```python
#  Q5: Radon transform using scipy.ndimage.rotate
def radon_rotate(image, theta, order=3):
    """Rotate the whole image with ndimage.rotate, then sum along the columns.
    The sign is -angle to reproduce the direction of skimage (image turned clockwise on screen)."""
    sinogram = np.zeros((image.shape[0], len(theta)))
    for i, angle in enumerate(theta):
        rotated = ndimage.rotate(image, -angle, reshape=False, order=order)   # order=3: scipy default
        sinogram[:, i] = rotated.sum(axis=0)
    return sinogram

# sign check: +angle would give the mirrored trajectories
report("radon_rotate(+angle) vs skimage", sino_sk, np.stack([ndimage.rotate(GT, a, reshape=False).sum(0) for a in THETA], 1))
report("radon_rotate(-angle) vs skimage", sino_sk, radon_rotate(GT, THETA))
```

**INTERPRETATIONS**

**5)** `radon_rotate` rotates the whole image with `ndimage.rotate(image, -angle, reshape=False)` and sums the columns. The sign must be **−θ** to reproduce skimage's direction (MSE 1.76 with −θ vs 11.4 with +θ, which gives mirrored sinusoids).

**6) Execution times.** skimage ≈ 40–60 ms; `radon_rotate` ≈ 180–200 ms with the default cubic spline (order 3, needs a spline prefilter of the whole image at each angle, ~3–5× skimage) and ≈ 65–80 ms with order 1 (same order as skimage, both being compiled bilinear warps); `radon_scratch` ≈ 35 s. Replacing the Python pixel loops by one compiled rotation per angle gives a ~200–500× speed-up: the cost becomes O(N² × N_angles) in C instead of Python.

**7) Comparison.** Sinogram SNR vs skimage = 20.4 dB (MSE 1.76): much better than `radon_scratch` (no holes) but still not identical. The difference map has a **sinusoidal, angle-dependent shift pattern** (positive on one side of each edge, negative on the other), and iradon(radon_rotate) has MSE 5.4e-3 (SNR 9.8 dB) vs 1.2e-3 (16.2 dB) for skimage: the edges are doubled/blurred. Why: **(a) rotation centre** — `ndimage.rotate` rotates about the true array centre (N−1)/2 = 49.5 while skimage rotates about N//2 = 50 (and `iradon` back-projects about 50). Rotating about 50 adds a translation of 0.5 px whose direction changes with θ, so the projections are inconsistent with skimage's geometry. Proof: the same column-sum with `affine_transform` about (50, 50) and bilinear interpolation reproduces skimage to **MSE 3e-28**. **(b) Interpolation** — `ndimage.rotate` uses cubic splines by default vs bilinear in `warp`: with the correct centre, order 3 still differs (MSE 3.6e-2, SNR 37 dB), a much smaller effect. (c) Boundary handling (`mode='constant'`, values outside set to 0) is identical here because the phantom is zero outside the disc. Conclusion: the main cause is the half-pixel centre convention, the interpolation order is secondary.

1. Use the given `inverse_radon()` function to reconstruct the sinogram. Looking at the code from the previous which step is missing in this FBP implementation?

**ANSWER:** `inverse_radon` only performs the **back-projection** (each projection is smeared along its rays with `np.tile`, rotated back to its angle and accumulated). The **filtering step is missing**: the projections are not convolved with the ramp filter (FFT → × |ω| → IFFT, lines 299–301 of `iradon`), so it is a simple (unfiltered) back-projection, not an FBP — the result is the image blurred by 1/r. The **normalisation** by π/(2·N_angles) (line 323) is also missing, so the amplitudes are wrong (see Q2).

```python
def inverse_radon(sinogram, theta):
    reconstruction = np.zeros((SIZE, SIZE), dtype=np.float64)
    stack = reconstruction
    for i, angle in enumerate(theta):
            proj_vect = sinogram[:, i]
            mat_proj = np.tile(proj_vect, (SIZE, 1))
            rot_pix = ndimage.rotate(mat_proj, angle, reshape=False)
            reconstruction = rot_pix
            stack = stack + reconstruction
    return stack
```

```python
def inverse_radon_norm(sinogram, theta):
    """Unfiltered back-projection + skimage normalisation pi / (2 * number_of_angles)  (iradon, line 323)"""
    size = sinogram.shape[0]
    reconstruction = np.zeros((size, size), dtype=np.float64)
    for i, angle in enumerate(theta):
        mat_proj = np.tile(sinogram[:, i], (size, 1))                    # smear the projection along the rays
        reconstruction += ndimage.rotate(mat_proj, angle, reshape=False)  # put it back at its angle
    return reconstruction * np.pi / (2 * len(theta))

bp_norm = inverse_radon_norm(sino_sk, THETA)
bp_norm_c = bp_norm * CIRCLE                                             # iradon also zeroes outside the disc
fig, ax = plt.subplots(1, 4, figsize=(18, 4))
for a, img, t in zip(ax[:3], [bp_norm, bp_sk, bp_norm_c - bp_sk],
                     ["inverse_radon_norm", "iradon(filter_name=None)", "difference (both masked to the disc)"]):
    lim = np.abs(img).max(); diff = "diff" in t
    h = a.imshow(img, cmap="seismic" if diff else "gray", vmin=-lim if diff else None, vmax=lim if diff else None)
    a.set_title(f"{t}\nrange [{img.min():.2f}, {img.max():.2f}]"); fig.colorbar(h, ax=a)
    a.set_xlabel("x (pixel)"); a.set_ylabel("y (pixel)")
for img, lab, sty in [(bp_sk, "iradon None", "b-"), (bp_norm_c, "inverse_radon_norm", "r--")]:
    ax[3].plot(profile_line(img, (5, 50), (95, 50), order=0), sty, label=lab)
ax[3].set_title("vertical profile, column 50"); ax[3].legend(); ax[3].grid(alpha=.3)
ax[3].set_xlabel("distance (pixel)"); ax[3].set_ylabel("intensity")
plt.tight_layout(); plt.show()
report("inverse_radon_norm vs iradon None (ref)", bp_sk, bp_norm_c)

# Same back-projection but rotating about skimage's centre (50) with bilinear interpolation
def backproject_exact(sinogram, theta):
    size = sinogram.shape[0]; rec = np.zeros((size, size))
    for i, angle in enumerate(theta):
        rec += rotate_skimage_centre(np.tile(sinogram[:, i], (size, 1)), -angle, order=1)
    return rec * np.pi / (2 * len(theta)) * CIRCLE
report("exact-centre back-projection vs iradon None", bp_sk, backproject_exact(sino_sk, THETA))
```

**ANSWERS:**

**1.** Normalise the back-projection as `iradon` does (line 323): `return reconstructed * np.pi / (2 * angles_count)`. The sum over N angles approximates ∫ dθ, so it must be multiplied by the angular step. Over [0, π[ with N angles Δθ = π/N. The figure above shows why the angular range matters: over 2π every line is measured twice (θ and θ + π give the same ray in opposite directions), so ∫₀^{2π} = 2∫₀^π and the 360° sum must be halved (Δθ/2 = π/N for N angles over 2π). skimage writes π/(2N) because its ramp filter is defined as 2|f| (`fourier_filter = 2 * np.real(fft(f))`), the factor 2 being compensated here; the unfiltered `iradon` keeps the same π/(2N).

**3.** After normalisation the ranges match ([−0.01, 29.0] for both) and the difference with `iradon(filter_name=None)` is small: SNR = 33.6 dB, MSE 0.16 (for values up to 29). The residual differences are located at the edges and come from the rotation centre (49.5 vs 50) and the interpolation (cubic spline 2D rotation vs 1D linear interpolation in `iradon`), plus the circle mask. Back-projecting around skimage's centre with bilinear interpolation (`backproject_exact`) reduces the difference to SNR 40.4 dB.

```python
def inverse_radon_fbp(sinogram, theta, filter_name="ramp", exact_centre=False):
    """Filtered back-projection written from the iradon steps:
       1) zero-pad each projection to a power of 2 >= 2N (avoids circular-convolution wrap-around)
       2) FFT along t, multiply by the filter, inverse FFT, crop   (iradon lines 299-301)
       3) back-project (smear + rotate) and normalise by pi/(2*N_angles) (iradon line 323)"""
    size = sinogram.shape[0]
    n_pad = max(64, int(2 ** np.ceil(np.log2(2 * size))))
    padded = np.pad(sinogram, ((0, n_pad - size), (0, 0)), mode="constant")
    fourier_filter = _get_fourier_filter(n_pad, filter_name)                 # shape (n_pad, 1)
    projection = fft(padded, axis=0) * fourier_filter
    sino_filtered = np.real(ifft(projection, axis=0)[:size, :])
    rec = np.zeros((size, size))
    for i, angle in enumerate(theta):
        smear = np.tile(sino_filtered[:, i], (size, 1))
        rec += (rotate_skimage_centre(smear, -angle, order=1) if exact_centre
                else ndimage.rotate(smear, angle, reshape=False))
    return rec * np.pi / (2 * len(theta)) * CIRCLE, sino_filtered

# 1) check against iradon (ramp)
rec_fbp, _ = inverse_radon_fbp(sino_sk, THETA, "ramp")
rec_fbp_exact, _ = inverse_radon_fbp(sino_sk, THETA, "ramp", exact_centre=True)
rec_iradon = iradon(sino_sk, THETA)
report("our FBP (ndimage.rotate) vs iradon", rec_iradon, rec_fbp)
report("our FBP (exact centre)   vs iradon", rec_iradon, rec_fbp_exact)
t_fbp = timeit.timeit(lambda: inverse_radon_fbp(sino_sk, THETA), number=3) / 3
t_irr = timeit.timeit(lambda: iradon(sino_sk, THETA), number=3) / 3
print(f"time: our FBP {t_fbp*1e3:.0f} ms | iradon {t_irr*1e3:.0f} ms")
```

### 1D approach

- The algebraic reconstruction technique (ART) is an iterative reconstruction technique used in computed tomography. It reconstructs an image from a series of angular projections (a sinogram).

- First we will understand the update of the coefficients in one dimension.
- Go to the https://scipython.com/blog/visualizing-kaczmarzs-algorithm/

- Iterativelly, we can solve a square matric which respects the invertibility theorem

```python
# code from https://scipython.com/blog/visualizing-kaczmarzs-algorithm/
A = np.array([[-4, 1], [2, 0.5], [3, 1.5], [0, 1]])
b = np.array([-2, 3, 6, 2.])
m, n = A.shape
A2 = np.array([[-4, 1], [2, 0.5], [3, 1.5], [0, 1]])
b2 = np.array([-2, 3, 4, 2.])
def solve_kaczmarz(A, b, TOL=1.e-6, randomize=False):
    def calc_err(X):
        return np.max(np.abs(A @ X - b))
    X = np.zeros(n)
    Xhistory = [X]
    err_history = [calc_err(X)]
    k = 0
    if randomize:
        p = (np.linalg.norm(A, axis=1) / np.linalg.norm(A))**2
    else:
        p = range(m)
    while True and k < 20:
        if randomize:
            i = np.random.choice(range(m), p=p)
        else:
            i = k % m
        ai = A[i,:]
        Xnew = X + (b[i] - ai @ X) / np.linalg.norm(ai)**2 * ai
        err = np.max(np.abs(A @ Xnew - b))
        Xhistory.append(Xnew)
        err_history.append(err)
        if err < TOL:
            break
        X = Xnew
        k += 1
    return X, np.array(Xhistory), np.array(err_history)

def plot_trajectory(Xhistory, A, b):
    x = np.linspace(-5, 5, 2)
    M = np.vstack((x, np.ones(2))).T
    for i in range(m):
        y = M @ (-A[i,0], b[i]) / A[i,1]
        plt.plot(x, y)
    plt.plot(Xhistory[:,0], Xhistory[:,1], 'k-o')
    plt.axis('square')
    plt.xlim(-0.5, 2.5)
    plt.ylim(-0.5, 2.5)
    plt.xlabel('x')
    plt.ylabel('y')

plt.figure(figsize=(30, 5))
plt.subplot(131)
X, Xhistory, _ = solve_kaczmarz(A, b, 1.e-2, randomize=False)
plot_trajectory(Xhistory, A, b)
print(f'Solution vector: {X}.')
plt.subplot(132)
X, Xhistory, _ = solve_kaczmarz(A, b, 1.e-2, randomize=True)
plot_trajectory(Xhistory, A, b)
print(f'Solution vector: {X}.')
plt.subplot(133)
X, Xhistory, _ = solve_kaczmarz(A2, b2, 1.e-2, randomize=False)
plot_trajectory(Xhistory, A2, b2)
print(f'Solution vector: {X}.')
plt.show()
```

**ANSWER (Kaczmarz).** Each equation aᵢ·x = bᵢ is a line (hyperplane) in the (x, y) plane. Kaczmarz's update **x_{k+1} = x_k + (bᵢ − aᵢ·x_k) / ‖aᵢ‖² · aᵢ** is the **orthogonal projection** of the current estimate onto the line of equation i: the correction is along the normal aᵢ and its size is the residual of that equation divided by ‖aᵢ‖². Cycling through the equations (left), the estimate zig-zags from line to line and converges to their intersection (1, 2) when the system is consistent. The randomized version (middle), choosing equation i with probability ∝ ‖aᵢ‖², converges at a rate that does not depend on the ordering of the equations (Strohmer & Vershynin, 2009); the result varies between runs because of the random choice. With an **inconsistent** system (right, b₃ changed so that the lines have no common point) it does not converge to a single point but cycles between the lines (limit cycle), stopped here by the 20-iteration limit: with noisy/inconsistent data a relaxation factor or an early stop is needed. ART applies exactly this to CT: one equation per ray, aᵢⱼ = length of ray i in pixel j, bᵢ = measured line integral.

### ART update one batch

#### Sources
- Introduction video : https://www.youtube.com/watch?v=hkJHL0DwUUQ&ab_channel=ASTRAToolbox
- Update pixels of the image : https://en.m.wikipedia.org/wiki/Algebraic_reconstruction_technique

```python
def ART_one_batch(sinogram, theta):
  size = sinogram.shape[0]
  X_rot = np.zeros((size, size))
  for a, angle in enumerate(theta):
    X_rot = ndimage.rotate(X_rot, - angle, reshape=False)
    S_calc = np.sum(X_rot, axis=0)
    diff = sinogram[:,a] - S_calc
    norm = np.linalg.norm(sinogram[:,a])

    for i in range(size):
        for j in range(size):
            X_rot[i,j] = X_rot[i,j] + diff[j]/norm
            if X_rot[i,j] < 0.01:
              X_rot[i,j] = 0
    X_rot = ndimage.rotate(X_rot, angle, reshape=False)
  return X_rot
```

**ANSWER (1. how `ART_one_batch` updates the pixels).** The image starts at 0. For each angle: (i) the current image is rotated by −θ (same direction as skimage) so that the rays of this projection become the columns; (ii) the computed projection `S_calc` is the sum of each column (forward projection); (iii) `diff = measured − computed` is the residual of each ray; (iv) the residual of ray j is added to **every pixel of column j** (back-projection of the error along the ray), divided by `norm` = ‖measured projection‖; (v) pixels < 0.01 are set to 0 (positivity constraint, also removes small values); (vi) the image is rotated back by +θ. It is one Kaczmarz/ART pass where all the rays of one angle are updated together (block ART, close to SART). Remarks: the classical step is diff / ‖aᵢ‖² = diff / (number of pixels on the ray); dividing by ‖p_θ‖ (≈ 150) is an arbitrary relaxation factor; the image itself is interpolated twice per angle (400 cubic rotations per pass), which progressively blurs it and rotates around (N−1)/2 instead of skimage's N/2; the double Python loop can be replaced by a broadcast (identical result, ~3× faster).

```python
# ---------- Q2 / Q3: ART one batch with an increasing number of projections + execution time ----------
proj_list = [10, 25, 50, 100, 150, 200]
art1, t_art = {}, {}
for n_p in proj_list:
    th = np.linspace(0, ANGLE, n_p)
    s = radon(GT, theta=th)
    t0 = time.perf_counter(); art1[n_p] = ART_one_batch(s, th); t_art[n_p] = time.perf_counter() - t0

fig, ax = plt.subplots(2, 4, figsize=(18, 8.5)); ax = ax.ravel()
h = ax[0].imshow(GT, vmin=0, vmax=1); ax[0].set_title("Ground truth"); fig.colorbar(h, ax=ax[0])
for a, n_p in zip(ax[1:], proj_list):
    h = a.imshow(art1[n_p], vmin=0, vmax=1); fig.colorbar(h, ax=a)
    a.set_title(f"ART one batch, {n_p} proj. ({t_art[n_p]:.2f} s)\nMSE={mse(GT, art1[n_p]):.2e}  SNR={snr_db(GT, art1[n_p]):.2f} dB")
ax[-1].plot(proj_list, [t_art[n] for n in proj_list], "o-"); ax[-1].set_xlabel("number of projections")
ax[-1].set_ylabel("execution time (s)"); ax[-1].set_title("ART_one_batch execution time"); ax[-1].grid(alpha=.3)
for a in ax[:-1]:
    a.set_xlabel("x (pixel)"); a.set_ylabel("y (pixel)")
plt.tight_layout(); plt.show()
for n_p in proj_list:
    print(f"{n_p:4d} proj: {t_art[n_p]:.3f} s  ({1e3*t_art[n_p]/n_p:.1f} ms / projection)  SNR={snr_db(GT, art1[n_p]):.2f} dB")

# the double pixel loop can be vectorised -> bit-identical result, much faster
def ART_one_batch_vec(sinogram, theta, X=None):
    size = sinogram.shape[0]
    X_rot = np.zeros((size, size)) if X is None else X.copy()
    for a, angle in enumerate(theta):
        X_rot = ndimage.rotate(X_rot, -angle, reshape=False)
        diff = sinogram[:, a] - X_rot.sum(axis=0)
        X_rot = X_rot + diff[None, :] / np.linalg.norm(sinogram[:, a])
        X_rot[X_rot < 0.01] = 0
        X_rot = ndimage.rotate(X_rot, angle, reshape=False)
    return X_rot
t0 = time.perf_counter(); art_vec = ART_one_batch_vec(sino_sk, THETA); t_vec = time.perf_counter() - t0
print(f"\nvectorised version (200 proj.): {t_vec:.3f} s vs {t_art[200]:.3f} s, identical: {np.allclose(art_vec, art1[200])}")
```

1. Modify the function to have multiple iterations with a stop condition on the plateau (you can use a criterion to monitor if the reocnstruction is better and stop when there is no change)

```python
#  ART with global iterations and a stopping rule on the plateau
def residual(X, sinogram, theta, projector):
    """data-consistency criterion (no ground truth needed):  ||p - R x|| / ||p||"""
    est = np.stack([projector(X, a).sum(axis=0) for a in theta], axis=1)
    return np.linalg.norm(sinogram - est) / np.linalg.norm(sinogram)

def ART(sinogram, theta, max_iter=15, tol=1e-2, relax=1.0, variant="improved", verbose=True):
    """variant='given'   : update rule of ART_one_batch (image rotated back and forth, step diff/||p||)
       variant='improved': Kaczmarz step with the ray length (diff / L), only the CORRECTION is back-rotated
                           (no cumulative interpolation of the image), projector = skimage geometry (centre N//2),
                           relaxation, positivity constraint, support = reconstruction disc.
       stop when the relative decrease of the residual between two global iterations < tol, or max_iter reached"""
    size = sinogram.shape[0]
    X = np.zeros((size, size))
    if variant == "given":
        projector = lambda img, a: ndimage.rotate(img, -a, reshape=False)
    else:
        projector = lambda img, a: rotate_skimage_centre(img, a, order=1)
        ray_len = {a: np.maximum(projector(CIRCLE.astype(float), a).sum(axis=0), 1) for a in theta}
    hist = dict(X=[], res=[], snr=[], time=[])
    res_prev = residual(X, sinogram, theta, projector)
    t0 = time.perf_counter()
    for it in range(1, max_iter + 1):
        if variant == "given":
            X = ART_one_batch_vec(sinogram, theta, X)
        else:
            for a, angle in enumerate(theta):
                diff = (sinogram[:, a] - projector(X, angle).sum(axis=0)) / ray_len[angle]
                X += relax * rotate_skimage_centre(np.tile(diff, (size, 1)), -angle, order=1) * CIRCLE
                np.maximum(X, 0, out=X)                                # positivity
        res = residual(X, sinogram, theta, projector)
        hist["X"].append(X.copy()); hist["res"].append(res); hist["snr"].append(snr_db(GT, X))
        hist["time"].append(time.perf_counter() - t0)
        gain = (res_prev - res) / res_prev
        if verbose:
            print(f"[{variant}] it {it:2d}: residual={res:.4f} (rel. decrease {100*gain:6.2f} %)  SNR={hist['snr'][-1]:.2f} dB")
        if gain < tol:
            print(f"[{variant}] stop: plateau reached (relative decrease < {100*tol:.0f} %) after {it} iterations")
            break
        res_prev = res
    else:
        print(f"[{variant}] stop: max_iter={max_iter} reached")
    return X, hist

X_given, h_given = ART(sino_sk, THETA, variant="given")
```

**ANSWER:**

**1. Iterations + stopping rule.** `ART()` repeats global passes over all the angles. Stopping criterion (no ground truth needed, as in practice): the **data residual** r_k = ‖p − R x_k‖ / ‖p‖; stop when its relative decrease between two global iterations is < 1 % (plateau) or when `max_iter` is reached. The SNR vs GT is only monitored.
- *Given update rule*: plateau after **3 iterations** (r = 0.128 → 0.123 → 0.123), SNR stuck at ≈ 9 dB: the cumulative interpolation blur and the half-pixel centre mismatch prevent the estimate from ever matching the data.
- *Improved ART*: classical Kaczmarz step diff / (ray length), only the correction is back-rotated (the image is never re-interpolated), projector identical to skimage's geometry (centre N//2, bilinear), positivity and disc support, relaxation 1. On the clean sinogram the residual decreases geometrically (0.08 → 1·10⁻⁴) and the SNR reaches 43.8 dB after 30 iterations, above SART 2 it. (21.5 dB) and FBP (16.2 dB); the plateau rule is not reached within 30 iterations, so `max_iter` stops it. **Caveat — inverse crime:** data and reconstruction use the same discrete projector and the data are noise-free, so this score is optimistic.
- *Noisy sinogram (σ = 2 % of max, relative noise 3.7 %)*: the residual plateaus at the noise level (0.041) and the rule stops after **6 iterations**; the SNR peaks at iteration 2 (15.45 dB) and then decreases (14.3 dB at it. 6): **semi-convergence** — the late iterations fit the noise. A better rule for noisy data is the discrepancy principle (stop when r_k ≤ τ·noise level) or relaxation < 1 / regularisation. FBP noisy 15.0 dB, SART 2 it. noisy 17.3 dB (its 0.15 relaxation acts as a regularisation).

**2. Profiles** (colour = iteration). Given rule: the profiles barely change after the 1st iteration and stay biased w.r.t. the GT: skull peaks only 0.55–0.8 instead of 1, brain plateau 0.12–0.25 instead of 0.20, large ellipse ≈ 0.25 instead of 0.30, wide edges. Improved ART: the first iterations are smooth (low frequencies first), then the edges steepen and the small ellipses appear; after a few iterations the profile is sharper than skimage's FBP and SART (steeper skull edges, correct plateaus). Noisy case: the edges sharpen during the first 2 iterations, then the plateaus become increasingly noisy — visual confirmation of the semi-convergence and of the need for an early stopping criterion.
