**2. Which field contains the actual image?**

The image is stored in **Pixel Data, tag (7FE0,0010)**, with VR `OW` (here 512 × 512 × 2 bytes = 524 288 bytes). `dataset.PixelData` gives the raw bytes. `dataset.pixel_array` decodes them into a NumPy array (512 × 512, `int16`) using the pixel-description fields (Rows, Columns, BitsAllocated, PixelRepresentation and the Transfer Syntax).

**3. What information is needed to visualise a DICOM image? What does *Pixel Spacing* mean?**

- **To decode the bytes:** the Transfer Syntax UID (byte order, implicit/explicit VR, compression such as JPEG, JPEG 2000 or RLE), Rows and Columns, Samples per Pixel, Bits Allocated/Stored, High Bit, Pixel Representation (signed or unsigned), and Number of Frames for multi-frame files.
- **To map values to grey levels:** Photometric Interpretation (MONOCHROME2 = low values black; MONOCHROME1 = inverted; RGB…), Rescale Slope and Intercept (stored value → modality value, e.g. Hounsfield units in CT), and Window Center/Width or a VOI LUT (which range of values is mapped onto the 0–255 range of the screen).
- **To show the correct geometry:** Pixel Spacing, plus Slice Thickness and Image Position/Orientation for 3D.

**Pixel Spacing (0028,0030)** is the physical distance in the patient, in mm, between the centres of two adjacent pixels. It is given as `[row spacing, column spacing]`: the first value is the vertical distance between adjacent rows, the second the horizontal distance between adjacent columns. Here it is 0.4492 mm × 0.4492 mm, so the pixels are square and the 512 × 512 image covers 230 mm × 230 mm. Pixel Spacing is what lets us:
- display the image with the correct aspect ratio when the two spacings differ;
- draw axes in mm (see the second figure below);
- measure real distances, areas and volumes, for example the size of a lesion in mm instead of in pixels.

### Investigate the image

To begin, check what range of the graylevels are present in the DICOM file for the image. Comment on what would be needed for visualization of such a range on our screens

```python
# ANSWER: grey-level range of the DICOM image
img = dataset.pixel_array
print("dtype                   :", img.dtype, "| BitsAllocated =", dataset.BitsAllocated,
      "| BitsStored =", dataset.BitsStored, "| PixelRepresentation =", dataset.PixelRepresentation, "(1 = signed)")
print("range allowed by dtype  :", np.iinfo(img.dtype).min, "to", np.iinfo(img.dtype).max)
print("range actually present  : min =", img.min(), ", max =", img.max())
print("number of distinct values:", np.unique(img).size)
print("fraction of zero pixels  : %.1f %%" % (100 * np.mean(img == 0)))
print("bits needed for max value:", int(np.ceil(np.log2(int(img.max()) + 1))))

fig, ax = plt.subplots(1, 2, figsize=(13, 4))
ax[0].hist(img.ravel(), bins=200, color="tab:blue")
ax[0].set_yscale("log"); ax[0].set_title("Histogram of all pixels (log scale)")
ax[0].set_xlabel("stored pixel value"); ax[0].set_ylabel("count")
ax[1].hist(img[img > 0].ravel(), bins=200, color="tab:orange")
ax[1].axvline(255, color="k", ls="--", label="255 = max of an 8-bit display")
ax[1].set_title("Histogram of non-zero pixels"); ax[1].set_xlabel("stored pixel value"); ax[1].set_ylabel("count")
ax[1].legend(); plt.tight_layout(); plt.show()
```

ANSWER

- **Storage:** the pixels are stored as signed 16-bit integers (`int16`; BitsAllocated = BitsStored = 16, PixelRepresentation = 1), so the format could hold values from −32768 to 32767.
- **Values present:** from **0 to 1335**, with 1274 distinct grey levels. About 69 % of the pixels are exactly 0 (air outside the head). The non-zero (head) values lie mainly between ~300 and ~1030 (5th–95th percentiles), with a median around 800.
- **Bit depth:** the value 1335 needs 11 bits, so the dynamic range is much larger than what an 8-bit screen can show.
- **Units:** MRI values are arbitrary units with no absolute physical meaning, unlike Hounsfield units in CT. They depend on the sequence, the coil and the scaling applied by the scanner.

**What visualisation requires.** A standard screen, like the PNG and JPEG formats, displays 8 bits per channel, i.e. **256 grey levels (0–255)**. Sending the raw values directly would saturate everything above 255 to white, and that is 96 % of the head pixels (dashed line on the histogram). A **grey-level (VOI) transformation that maps the useful range onto [0, 255]** is therefore needed. Three options:
1. **Linear Min-Max rescaling** of [min, max] onto [0, 255] (next cells). It is simple and keeps every value, but 1274 levels are compressed into 256 and subtle contrasts are lost when the result is quantised.
2. **Windowing** with Window Center/Width (header fields, used in Lab 4). Only the range of interest is spread over the 256 levels; values outside the window are clipped to black or white.
3. **Non-linear mappings** such as histogram equalisation, gamma correction or a lookup table (LUT).

`plt.imshow` already applies an implicit linear Min-Max (vmin = min, vmax = max) when no limits are given, which is why the image could be displayed directly above. Diagnostic monitors use 10–12 bits (1024–4096 levels) for the same reason.

There are multiple ways of handling this. What is the maximum value of graylevel that we can display on our screen? For the rest of the assignment we denote this as **MAXVAL**. Thus we want the graylevels to be in the range [0, **MAXVAL**]


**ANSWER:** An 8-bit display has $2^8 = 256$ grey levels per channel, numbered 0 to 255. Therefore **MAXVAL = 255**, and the target range is **[0, 255]**. Medical-grade monitors can display 10 bits (MAXVAL = 1023) or more, but 255 is the standard value and is used in the rest of this assignment.

ANSWERS :

1. **ANSWER – Objective function.** In optimisation, the objective function (also called cost or energy function) is the scalar function $E(u)$ that scores a candidate solution $u$. The method searches for the solution that minimises it: $\hat u = \arg\min_u E(u)$. In variational denoising, $E$ has a **data-fidelity term** and a **regularisation term**:
$$E(u) = \underbrace{\tfrac{1}{2}\|u - f\|_2^2}_{\text{fidelity: stay close to the noisy data } f} \;+\; \underbrace{\mu\, R(u)}_{\text{regularisation: prior on what a clean image looks like}}$$
The parameter $\mu$ (`weight` in skimage) balances the two terms. The fidelity term alone is minimised by the noisy image itself ($u = f$), the regularisation term alone by a constant image, and the solution lies between the two.

2. **ANSWER – Total-variation regularisation.** The total variation measures how much an image varies in total. It is the integral of the norm of the image gradient (a sum over pixels in the discrete case):
$$TV(u)=\int_\Omega |\nabla u|\,dx \qquad\text{discrete: } TV(u)=\sum_{i,j}\sqrt{(u_{i+1,j}-u_{i,j})^2+(u_{i,j+1}-u_{i,j})^2}$$
TV denoising (the Rudin–Osher–Fatemi, ROF, model) solves
$$\hat u=\arg\min_u\ \tfrac{1}{2}\sum_i (u_i-f_i)^2 + \text{weight}\cdot TV(u).$$
This is the form in the skimage documentation, $\min_u \sum_i |\nabla u_i| + \frac{\lambda}{2}(f_i-u_i)^2$ with $\lambda = 1/\text{weight}$, multiplied by the constant weight. Noise creates many small, random oscillations, and each one adds to the sum of gradient magnitudes, so a noisy image has a large TV. Minimising $E$ removes these oscillations, while the fidelity term keeps the result close to the data. Because TV is the **L1 norm of the gradient**, the solution is piecewise smooth (almost piecewise constant) with sharp edges.

3. **ANSWER – Why TV preserves shapes, compared with L² regularisation.**
   - **L² (Tikhonov / Sobolev) regularisation**, $R(u)=\int|\nabla u|^2$, penalises gradients **quadratically**. A jump of height $h$ concentrated on one pixel costs about $h^2$, whereas the same jump spread over $n$ pixels costs only $n\,(h/n)^2 = h^2/n$. The minimiser therefore always spreads edges out, i.e. blurs them. Its Euler–Lagrange equation, $u - f = \mu\,\Delta u$, is a linear, heat-equation-like diffusion that acts as a low-pass filter. Edges and noise are both high frequencies, so they are removed together.
   - **TV**, $R(u)=\int|\nabla u|$ (the L1 norm of the gradient), penalises gradients **linearly**. In 1D, a monotone transition of height $h$ costs exactly $|h|$, whether it is a sharp step or a smooth ramp, so there is no incentive to smear a real edge. Noise, in contrast, goes up and down many times, and every oscillation adds to the TV, so it is penalised heavily relative to its small amplitude.
   - The TV Euler–Lagrange equation, $u - f = \mu\,\mathrm{div}\!\left(\nabla u/|\nabla u|\right)$, is a **non-linear diffusion** with coefficient $1/|\nabla u|$. It smooths strongly in flat regions (small gradient) and acts only along strong edges, not across them. By the co-area formula, TV is also the integral of the perimeters of the level sets: it penalises the length of the boundaries, not the contrast across them.
   - As with the L1 norm in sparse regression (LASSO), TV favours a **sparse gradient**: most gradients become exactly zero (flat regions) and a few large jumps (the edges) are kept. Shapes are therefore preserved.
   - The price, visible in our results, is the staircasing (cartoon) effect, the loss of fine texture, and a loss of contrast in small objects (proportional to weight × perimeter / area).

4. **ANSWER – Can we directly use gradient descent? No.**
   - The TV term $\sum_i |\nabla u_i|$ is convex but **not differentiable**: the Euclidean norm has no derivative at 0. Its "gradient" $-\mathrm{div}(\nabla u/|\nabla u|)$ is undefined (division by zero) wherever $\nabla u = 0$, which is exactly the flat regions that the TV solution creates. Plain gradient descent has no gradient to follow there.
   - The workarounds are poor. **Subgradient descent** converges slowly ($O(1/\sqrt{k})$), needs decreasing step sizes and oscillates around the kinks. **Smoothing the TV**, e.g. $\sqrt{|\nabla u|^2+\epsilon^2}$, makes it differentiable, but its gradient has a Lipschitz constant of order $1/\epsilon$. Stable step sizes must then be tiny (very slow convergence), and the result is only an approximation: gradients are never exactly zero, so edges are slightly blurred.
   - **Primal–dual and dual methods** avoid the problem by writing TV in its dual form, $TV(u)=\max_{|p_i|\le 1}\ \langle \nabla u, p\rangle = \max_{|p_i|\le 1}\ \langle u, -\mathrm{div}\,p\rangle$. The non-smooth term becomes a smooth bilinear term plus a simple constraint, and the projection onto that constraint is trivial. Each iteration only needs gradients, divergences and projections, with guaranteed convergence and a fixed step size. Chambolle–Pock (2011) alternates primal and dual steps. The function `denoise_tv_chambolle` used here implements the closely related Chambolle (2004) dual projection algorithm (its reference [1]). It iterates on the dual field $p$ with step $\tau = 1/4$ and returns $u = f - \mathrm{div}\,p$, where $|p_i| \le$ weight.

5. (OPTIONAL) **ANSWER (BONUS) – Isotropic or anisotropic?** **Isotropic.**
   - In the skimage source (`_denoise_tv_chambolle_nd`), the gradient is computed with forward differences along each axis (`np.diff`, with zero difference at the last row/column, i.e. Neumann boundary conditions).
   - The per-pixel norm is the **Euclidean norm of the whole gradient vector**, `norm = np.sqrt((g**2).sum(axis=0))`, so the regulariser is
   $$TV_{iso}(u)=\sum_{i,j}\sqrt{(\partial_x u)_{i,j}^2+(\partial_y u)_{i,j}^2}.$$
   This is exactly the discrete total variation $J(u)=\sum_{i,j}|(\nabla u)_{i,j}|$ defined by Chambolle (2004).
   - The **anisotropic** TV would use the L1 norm of the gradient, $\sum_{i,j}|(\partial_x u)_{i,j}|+|(\partial_y u)_{i,j}|$. It is separable along the axes and favours edges aligned with the grid (horizontal or vertical), giving blocky results. The isotropic version couples the partial derivatives, so its penalty depends much less on the orientation of an edge.
   - Two remarks. First, "isotropic" refers to the form of the norm: on a discrete grid with one-sided differences, this TV is still not exactly rotation-invariant (a slight directional bias remains). Second, in nD the same Euclidean norm is taken over all spatial axes; with `channel_axis`, each channel is denoised independently, so there is no vectorial (colour) TV.

ANSWERS:

1. **ANSWER – Parameters `patch_size`, `patch_distance`, `h` and `sigma`.** NLM replaces each pixel $i$ by a weighted average of the pixels $j$ of its search window:
$$\hat u(i)=\frac{1}{C(i)}\sum_{j} w(i,j)\,f(j),\qquad w(i,j)=\exp\!\left(-\frac{\max\!\big(d^2(P_i,P_j)-2\sigma^2,\;0\big)}{h^2}\right),\qquad C(i)=\sum_j w(i,j).$$
Here $d^2$ is the mean squared difference between the patches $P_i$ and $P_j$ centred on $i$ and $j$. In `fast_mode` every pixel of the patch counts equally; the original algorithm uses Gaussian weights inside the patch.
   - **`patch_size`**: side length of the square patches that are compared (default 7, i.e. 7 × 7). Larger patches give a more robust similarity measure under strong noise (more pixels in $d^2$) and compare real structures (edges, textures) rather than single pixels. But truly similar patches become rarer, so small details may be averaged with different structures, and the classic algorithm slows down (cost ∝ patch_size², almost independent of it in fast mode). Patches that are too small (3) give a similarity dominated by noise.
   - **`patch_distance`**: maximum distance, in pixels, at which candidate patches are searched. The search window therefore has $(2d+1)^2$ positions, with $d$ = `patch_distance` (default 11, i.e. 23 × 23). A larger window finds more similar patches and averages more, giving stronger denoising. However, the cost grows as patch_distance², and many dissimilar patches add small weights that blur the image. A truly global search would be far too expensive, so in practice "non-local" means "semi-local".
   - **`h`**: cut-off, or filtering parameter, in grey levels. It sets how fast the weights decay with the patch distance. With a small h, only nearly identical patches contribute: the result stays close to the input and noise remains. With a large h, dissimilar patches are also averaged: smoothing is stronger and edges and fine details blur. h should be of the order of the noise σ (≈ σ without `sigma`; ≈ 0.8σ in fast mode with `sigma`).
   - **`sigma`**: standard deviation of the (Gaussian) noise, 0 by default. Two noisy patches with the same clean content still differ, by $2\sigma^2$ on average. When `sigma` is given, this expected value is subtracted from $d^2$ (the max(·, 0) term above), so patches whose difference is explained by noise alone get a weight close to 1. This makes the weights more robust and gives a moderate gain, provided that a smaller h is used. A wrong σ, such as our underestimated σ̂ from `estimate_sigma` (wavelet MAD estimator), reduces the benefit.

2. **ANSWER – What if `patch_distance` < `patch_size`?** `patch_size` is the full width of a patch, whereas `patch_distance` is a radius: the search window is $2d+1$ pixels wide (with $d$ = `patch_distance`). If `patch_distance` < `patch_size`:
   - The search window is small and contains only $(2d+1)^2$ candidates (9 for d = 1, against 529 for d = 11). Few pixels are averaged, so the **noise reduction is weak**.
   - Every candidate is shifted by less than one patch width, so it **overlaps strongly with the reference patch** and shares most of its pixels, including their noise. The distances are artificially small and correlated, so similarity is no longer measured on independent data.
   - The method **loses its non-local nature**: it cannot use similar structures elsewhere in the image and becomes a small local adaptive filter, close to a bilateral or sigma filter.
   - In the limit `patch_distance = 0`, the only candidate is the pixel itself, so the output equals the input. skimage raises no error; it simply denoises less.

   The experiment above (patch_size = 7) confirms this:
   - d = 0 returns the noisy image unchanged (PSNR 18.19 dB, max |output − input| = 0);
   - d = 1 and d = 2 give only 22.64 dB and 23.96 dB;
   - quality approaches its maximum only when d ≥ 5, i.e. close to the patch size (24.57 dB at d = 5–6);
   - beyond that it decreases slightly (24.25 dB at d = 20), while the run time grows quadratically (0.02 s → 2.4 s).
