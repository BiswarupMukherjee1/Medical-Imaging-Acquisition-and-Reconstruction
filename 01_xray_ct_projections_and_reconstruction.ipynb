```python
# Imports used in the whole notebook
import time
import warnings
import numpy as np
import matplotlib.pyplot as plt
from skimage.data import shepp_logan_phantom
from skimage.transform import radon, iradon, iradon_sart, resize
from skimage.transform.radon_transform import _get_fourier_filter
from skimage.measure import profile_line
from skimage.metrics import structural_similarity

plt.rcParams.update({"image.cmap": "gray", "figure.dpi": 90, "axes.titlesize": 10})
```

```python
# Q1: load the Shepp-Logan phantom and resize it
SIZE, PROJ, ANGLE = 300, 1000, 360                      # parameters imposed for the rest of the lab
SL_full = shepp_logan_phantom()                          # 400 x 400, float64 in [0, 1]
GT = resize(SL_full, (SIZE, SIZE), anti_aliasing=True)   # ground truth used everywhere below
THETA = np.linspace(0., ANGLE, PROJ, endpoint=False)     # 1000 angles on [0, 360[, step 0.36 deg

print("original:", SL_full.shape, SL_full.dtype, "| resized:", GT.shape,
      "| grey levels of the original phantom:", np.unique(np.round(SL_full, 2)))

fig, ax = plt.subplots(1, 2, figsize=(10, 4.3))
for a, im, t in zip(ax, [SL_full, GT], ["Shepp-Logan phantom (skimage, 400x400)",
                                       f"Resized phantom = ground truth ({SIZE}x{SIZE})"]):
    h = a.imshow(im, vmin=0, vmax=1); a.set_title(t)
    a.set_xlabel("x (column, pixel)"); a.set_ylabel("y (row, pixel)")
    fig.colorbar(h, ax=a, label="attenuation (a.u.)")
plt.tight_layout(); plt.show()
```

```python
#  Q2: rotation direction of skimage.radon (numerical check of the rotation matrix)
# In radon():  R = [[cos a,  sin a, ...],
#                   [-sin a, cos a, ...]]  is passed to warp() as the *inverse* map (output -> input coords)
# Test: a single bright point placed ABOVE the centre, and one to the RIGHT of the centre.
n = 41; c = n // 2
pt_up = np.zeros((n, n));    pt_up[c - 10, c] = 1      # 10 px above centre
pt_right = np.zeros((n, n)); pt_right[c, c + 10] = 1   # 10 px right of centre
angles = [0, 45, 90, 135, 180]
s_up, s_right = radon(pt_up, angles), radon(pt_right, angles)
for a, i, j in zip(angles, s_up.argmax(0), s_right.argmax(0)):
    print(f"theta={a:3d} deg | point above centre -> detector bin {i:2d} (t={i-c:+3d})"
          f" | point right of centre -> bin {j:2d} (t={j-c:+3d})")
print("=> t(theta) = x.cos(theta) + y.sin(theta) with y pointing UP: the point at the top goes to +t at 90 deg,")
print("   i.e. the image is turned CLOCKWISE on screen before the column sums (= source/detector turning counter-clockwise).")
```

**INTERPRETATIONS**

**1) Phantom.** A phantom is a test object whose exact attenuation map is known, so a reconstruction can be judged against a true reference (ground truth), which is impossible on a patient. The Shepp–Logan phantom (Shepp & Logan, 1974) is a 2D head model made of 10 ellipses: a dense skull ring (1.0), brain (0.2), two dark ventricles (0.0), a large ellipse (0.3) and small "tumours" (0.3–0.4). It is special because (i) it is defined analytically, so its Radon transform is known in closed form and errors can be attributed to the algorithm rather than to the data; (ii) it combines a very high-contrast structure (skull) with low-contrast details (0.1 contrast) and very small objects (3 ellipses of ~7 px), so it tests at once contrast resolution, spatial resolution and artefacts (streaks, ringing near the skull). skimage uses the *modified* version (Toft) with more visible contrasts: grey levels {0, 0.1, 0.2, 0.3, 0.4, 1}.

**2) Sinogram and direction of rotation.** A sinogram is the image of all the projections: column θ contains the line integrals ∫μ dl of the object along parallel rays at angle θ, row t is the position of the ray on the detector. Every point of the object at polar coordinates (r, φ) is projected at t = r·cos(θ − φ), so it draws a sinusoid whose amplitude is its distance to the rotation centre and whose phase is its angle — hence the name. In `radon`, the matrix `R = [[cos θ, sin θ], [−sin θ, cos θ]]` is given to `warp` as the inverse mapping (output → input). The numerical test above shows that a point placed above the centre is projected at t = +10 for θ = 90°, i.e. t = x cos θ + y sin θ with y pointing up: the image is turned **clockwise** on screen before the column sum, which is equivalent to the source/detector pair turning **counter-clockwise** (θ measured counter-clockwise from the x-axis, Kak & Slaney convention).

**3) Grey levels.** The phantom stores the attenuation μ ∈ [0, 1] of each pixel, whereas the sinogram stores *sums* of μ along a ray (units μ × pixel). A ray crosses up to ~250 pixels, so values reach 78.2. They are highest where a ray crosses the skull tangentially (long path in dense bone, bright sinusoidal edges) and they depend on the image size. Each projection preserves the total mass: the mean column sum (11088.4) equals the sum of the image.

**4) Property.** Changing the number of projections only changes the angular sampling (number of columns), not the values. Over 360° the second half of the sinogram is the first half flipped along t: **R(t, θ + 180°) = R(−t, θ)** (relative difference 2.6e-15). A parallel-beam acquisition over 180° therefore contains all the information; 360° is redundant (periodicity 2π, symmetry π). Other properties visible: linearity, mass conservation per projection, and every point mapping to a sinusoid.

```python
#  Q1: default filter of iradon
import inspect
print("iradon default filter_name =", inspect.signature(iradon).parameters["filter_name"].default)
f_ramp = _get_fourier_filter(512, "ramp")[:, 0]
freq = np.fft.fftfreq(512)
plt.figure(figsize=(5, 3))
plt.plot(np.fft.fftshift(freq), np.fft.fftshift(f_ramp), label="skimage 'ramp' (Ram-Lak)")
plt.plot(np.fft.fftshift(freq), 2 * np.abs(np.fft.fftshift(freq)), "--", label="ideal $2|f|$")
plt.xlabel("frequency (cycles / detector bin)"); plt.ylabel("gain"); plt.title("Default FBP filter: ramp")
plt.legend(); plt.grid(alpha=.3); plt.tight_layout(); plt.show()
print(f"gain at DC = {f_ramp[0]:.4f} (not exactly 0: ramp built in the spatial domain, Kak & Slaney eq. 61)")
```

```python
# Metrics used for the whole lab
def mse(ref, img):
    """Mean Square Error  MSE = 1/N * sum (ref - img)^2"""
    return np.mean((ref - img) ** 2)

def snr_db(ref, img):
    """Signal-to-Noise Ratio  SNR = 10 log10( sum ref^2 / sum (ref - img)^2 )   [dB]"""
    return 10 * np.log10(np.sum(ref ** 2) / np.sum((ref - img) ** 2))

def ssim(ref, img):
    return structural_similarity(ref, img, data_range=ref.max() - ref.min())

# ---------- Q2: progressive number of projections ----------
n_list = [8, 16, 32, 64, 128, 256, 512, 1000]
recs = {}
for n_p in n_list:
    th = np.linspace(0., ANGLE, n_p, endpoint=False)
    recs[n_p] = iradon(radon(GT, theta=th), theta=th)     # default ramp filter

fig, ax = plt.subplots(3, 3, figsize=(12, 12))
ax = ax.ravel()
h = ax[0].imshow(GT, vmin=0, vmax=1); ax[0].set_title("Original phantom (GT)")
for a, n_p in zip(ax[1:], n_list):
    a.imshow(recs[n_p], vmin=0, vmax=1)
    a.set_title(f"FBP, {n_p} proj.  MSE={mse(GT, recs[n_p]):.1e}\nSNR={snr_db(GT, recs[n_p]):.1f} dB  SSIM={ssim(GT, recs[n_p]):.2f}")
for a in ax:
    a.set_xlabel("x (pixel)"); a.set_ylabel("y (pixel)")
fig.colorbar(h, ax=ax.tolist(), shrink=0.6, label="attenuation (a.u.), same scale [0,1] for all")
plt.suptitle("iradon (ramp FBP) with an increasing number of projections over 360$^\\circ$", y=0.92)
plt.show()

fig, ax = plt.subplots(1, 3, figsize=(13, 3.2))
ax[0].semilogx(n_list, [mse(GT, recs[n]) for n in n_list], "o-"); ax[0].set_ylabel("MSE")
ax[1].semilogx(n_list, [snr_db(GT, recs[n]) for n in n_list], "o-"); ax[1].set_ylabel("SNR (dB)")
ax[2].semilogx(n_list, [ssim(GT, recs[n]) for n in n_list], "o-"); ax[2].set_ylabel("SSIM")
for a in ax:
    a.set_xlabel("number of projections"); a.grid(alpha=.3)
plt.suptitle("Global criteria vs number of projections"); plt.tight_layout(); plt.show()
n_nyq = int(np.ceil(np.pi * SIZE / 2))
print(f"Angular Nyquist: ~ pi*N/2 = {n_nyq} views over 180 deg (= {2*n_nyq} over 360 deg) for N={SIZE}")
```

**ANSWER**

**1. Default filter:** `filter_name='ramp'` (Ram-Lak). Back-projection alone over-weights low frequencies: by the Fourier-slice theorem the projections fill the 2D Fourier space on a polar grid whose sampling density decreases as 1/|ω|. The ramp filter multiplies each projection spectrum by |ω| to compensate this density (it is the Jacobian of the polar change of variables). skimage builds it in the spatial domain and then takes its FFT (Kak & Slaney, eq. 61), which avoids the DC offset of a frequency-sampled ramp (gain 0.0008 at DC instead of exactly 0) and stores it as 2|f| (normalisation compensated by π/(2N) at the end of `iradon`). It amplifies high frequencies, hence also noise.

**2. Number of projections.** With few projections the image is dominated by streak artefacts: each projection is back-projected as a band, and with too few angles the bands do not cancel (8 projections over 360° are only 4 distinct directions because θ and θ+180° are redundant). SNR increases from −6.5 dB (8 proj.) to 15.0 dB (128) and 20.2 dB (1000), SSIM from 0.20 to 0.97. The curve saturates beyond ~500 projections: the angular Nyquist criterion requires ≈ πN/2 = 472 views over 180°, i.e. ≈ 944 over 360°, so 1000 projections are just above the limit; the remaining error (≈ 20 dB) comes from discretisation and the ramp filter at the skull edges, not from angular undersampling.

**1. Filters.** skimage proposes ramp, Shepp-Logan, cosine, Hamming and Hann. All start from the ramp |ω| and multiply it by a low-pass window: Shepp-Logan by sinc(ω/2) (gentle roll-off, 0.64 at Nyquist), cosine by cos (0 at Nyquist), Hamming and Hann by raised-cosine windows (strongest attenuation). On this noise-free sinogram the ramp gives the lowest error (SNR 20.22 dB), then Shepp-Logan 19.12, cosine 17.25, Hamming 16.13, Hann 15.83 dB: the more the window attenuates high frequencies, the more the edges are blurred (profile through the small ellipses: the dips between them disappear with Hann). On noisy clinical data the trade-off reverses: apodised filters reduce high-frequency noise and streaks at the cost of resolution, which is why they exist (ramp = sharp/noisy, Hann = smooth).

**2. Without filter.** The image is a very blurred version of the phantom with a huge offset: values up to 87 instead of 1, structures barely visible, ventricles washed out (MSE = 3.4e3). Simple back-projection smears every projection back over the whole image, so every point contributes to all the others. The point-spread function computed above is exactly **1/(2r)**: unfiltered back-projection = true image ∗ 1/r, i.e. in Fourier space the image spectrum is multiplied by 1/|ω| (low frequencies over-weighted, high frequencies = edges attenuated). The filtering step is therefore essential in FBP: the ramp |ω| exactly cancels this 1/|ω| (Fourier-slice theorem, polar sampling density), restoring the correct amplitude and the sharp edges; the apodisation windows then control the noise/resolution trade-off.

### Bonus 2 : follow the Radon transform of a point with a sinus function

You will try to follow the sinus curve in the sinogram.
Using the link between the cartesian space and the polar space, having (x, y) coordinates you can find the amplitude and phase of the sinus.
Compute the function `polar_transform`

```python
w = 15, 10
x_new, y_new = w                     # point at column x=15, row y=10
size = 40
proj = 100
img = np.zeros((size, size))
img[y_new, x_new] = 1
S = radon(img, theta=np.linspace(0, 360, proj))
offset = size//2
f = 1/(proj - 1)   # template used 1/proj; theta includes the 360 deg endpoint, so the exact angular step is 1/(proj-1) of a turn
proj_vect = np.float64([i for i in range(proj)])

def sinusoidal_curve(x, amplitude, phase):
    return amplitude * np.sin(2 * np.pi * f * x + phase) + offset

def polar_transform(x, y, offset):
    # centred cartesian coordinates, y axis pointing UP (rows point down)
    dx, dy = x - offset, offset - y
    A = np.hypot(dx, dy)                 # amplitude = distance to the rotation centre
    phi0 = np.arctan2(dy, dx)            # polar angle of the point
    # skimage: t(theta) = dx cos(theta) + dy sin(theta) = A cos(theta - phi0) = A sin(theta - phi0 + pi/2)
    phi = np.pi / 2 - phi0
    return A, phi

plt.figure(figsize=(30, 5))
plt.subplot(121)
plt.imshow(img, cmap="gray")
plt.plot(x_new, y_new, "*y", label="Point")
plt.plot(offset, offset, "*r", label="Rotation center")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Image of a point at (%.2i, %.2i)" %(x_new, y_new))
plt.legend()

plt.subplot(122)
plt.plot([0, proj_vect[-1]], [offset, offset], "--r", label="Rotation center")
polar_coord = polar_transform(x_new, y_new, offset)
plt.plot(proj_vect, sinusoidal_curve(proj_vect, polar_coord[0], polar_coord[1]), "--", label="Polar coordinates (%.2f, %.2f)" %(polar_coord[0], polar_coord[1]))
plt.imshow(S, cmap="gray")
plt.legend()
plt.title("Sinogram")
plt.xlabel("Projections")
plt.show()# x_new, y_ne

peak = S.argmax(axis=0)                                  # measured trace of the point
model = sinusoidal_curve(proj_vect, *polar_coord)
print(f"A = {polar_coord[0]:.2f} px, phase = {np.rad2deg(polar_coord[1]):.1f} deg ; "
      f"RMS(model - measured peak row) = {np.sqrt(np.mean((model - peak)**2)):.2f} px (< 1 detector bin)")
```

**ANSWER (Bonus 2).** With centred coordinates dx = x − c, dy = c − y (y up), skimage projects the point at t(θ) = dx·cos θ + dy·sin θ = A·cos(θ − φ₀) with **A = √(dx² + dy²)** (distance to the rotation centre) and φ₀ = atan2(dy, dx) (its polar angle). Writing it as a sine, A sin(θ + φ) with **φ = π/2 − φ₀**. For the point (15, 10): A = 11.18 px, φ = −26.6°. The fitted curve follows the sinogram trace with an RMS error of 0.27 px (< 1 bin, due to pixel discretisation). The frequency was set to 1/(proj − 1) because `np.linspace(0, 360, proj)` includes the 360° endpoint.

**INTERPRETATIONS**

**1) SART** (Simultaneous Algebraic Reconstruction Technique, Andersen & Kak 1984) is an **iterative algebraic** reconstruction. The problem is written as a linear system **A x = p** (x = pixels, p = sinogram, a_ij = contribution of pixel j to ray i). Starting from an image (0 by default), for each projection angle it (a) forward-projects the current estimate, (b) computes the difference with the measured projection, (c) back-projects this residual normalised by the ray lengths, and (d) adds it with a relaxation factor (0.15 in skimage), all pixels being updated simultaneously from all the rays of one projection (vs. ray by ray in ART). One "iteration" = one pass over all the angles. It handles incomplete/noisy data and constraints (positivity) better than FBP but is much slower (≈ 5 s vs 0.8 s here). After 1 iteration: MSE = 3.5e-4, SNR = 22.2 dB, already better than FBP (20.2 dB).

**2)** `iradon_sart(..., image=previous)` restarts from the previous estimate. **Important:** skimage updates `image` **in place** (`image += relaxation * update` on `np.asarray(image)`), so the previous result must be passed as a copy, otherwise `sart[1]` is overwritten and seems identical to `sart[2]`. With the copy: 2 iterations → SNR 23.8 dB.

**3) Criteria.** SNR = 10·log₁₀(Σ GT² / Σ (GT − rec)²) [dB]; CNR = |μ_ROI − μ_BG| / σ_BG (ROI = ellipse p7, GT 0.3; BG = homogeneous brain, GT 0.2). The SNR increases monotonically with iterations (22.2 → 26.1 dB after 10, MSE 3.5e-4 → 1.4e-4) with decreasing gains (+1.5 dB, then +0.15 dB/iteration): the method converges towards the least-squares solution and saturates. The CNR **decreases** (24.4 → 11.3): the first iterations recover low frequencies (smooth image, low σ_BG) and the later ones add high frequencies — sharper edges, but also ripple/texture in flat regions. This is the classic bias/variance trade-off of iterative methods: with noisy data the iterations would amplify noise, so the number of iterations acts as a regularisation parameter.

**4) SART (3 it.) vs FBP vs phantom.** Globally SART is better: MSE 2.1e-4 vs 5.6e-4 (SNR 24.4 vs 20.2 dB). The error maps explain why: FBP's error is concentrated on the skull edges (smoothing + ringing of the ramp filter), whereas SART reproduces the sharp high-contrast edges better. However SART has a fine high-frequency texture inside the brain and outside the skull (σ_BG = 0.0068 vs 0.0023 for FBP), giving a lower SSIM (0.953 vs 0.974) and CNR (14.5 vs 42). Conclusion: SART is more accurate on average and on edges, FBP is smoother in homogeneous regions and 6× faster; the "best" method depends on the criterion — which shows that one global metric is not enough.

**BONUS — local vs global criteria**

*Global criteria* (MSE, SNR/PSNR, SSIM) summarise the whole image in one number: objective, easy to compare and to track over iterations. Limitations: MSE/SNR need a ground truth (only phantoms/simulations); they are dominated by large high-contrast errors (here the skull ring) and blind to *where* the error is — a missed 7-pixel tumour barely changes the MSE; a global intensity shift or a 1-pixel misregistration gives a large MSE although the image is diagnostically perfect, while a blurred image can have a better MSE than a sharp but slightly noisy one; SSIM depends on its window and data range. CNR does not need a ground truth but depends on the arbitrary choice of ROI/background and explodes when σ_BG → 0 (noise-free or over-smoothed images: FBP has CNR 42 here only because its background is very flat).

*Local criteria* (profiles, MTF, ROI statistics) describe resolution, edge sharpness, ringing and detectability of specific structures — what matters clinically — but are sensitive to the chosen line/ROI and are hard to rank automatically. In practice both are combined: a global metric to compare methods, local ones to explain the differences (as in the SART vs FBP comparison above, where MSE favours SART while CNR and SSIM favour FBP).
